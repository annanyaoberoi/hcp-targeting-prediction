```python
import requests

url = "https://data.cms.gov/data-api/v1/dataset/2a6705e6-7a1e-460c-ba22-35249a531918/data"

response = requests.get(
    url,
    params={"size": 5, "offset": 0},
    timeout=30
)

print("Status code:", response.status_code)

if response.status_code == 200:
    data = response.json()
    print("Data retrieved successfully!")
    print(data)
else:
    print("Request failed.")
    print(response.text[:500])
```

```python
import requests
import pandas as pd
import time

url = "https://data.cms.gov/data-api/v1/dataset/f68114ed-f854-4ffc-9c6e-ed78b5e2f8d0/data"

all_rows = []
offset = 0
batch_size = 5000

while True:
    response = requests.get(
        url,
        params={
            "filter[Brnd_Name]": "Jardiance",
            "size": batch_size,
            "offset": offset
        },
        timeout=60
    )

    response.raise_for_status()
    rows = response.json()

    if not rows:
        break

    all_rows.extend(rows)
    offset += len(rows)

    print("Rows retrieved so far:", len(all_rows))

    if len(rows) < batch_size:
        break

    time.sleep(0.2)

df_jardiance = pd.DataFrame(all_rows)

df_jardiance.to_csv("jardiance_2021.csv", index=False)

print("\nTotal rows:", len(df_jardiance))
print("Columns:", df_jardiance.columns.tolist())

display(df_jardiance.head())
```
